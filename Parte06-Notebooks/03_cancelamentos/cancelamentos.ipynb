# Análise de dataset


A empresa Umbrella Corporation contratou desenvolvedores Python (🐍) para analisar um dataset para verificar o alto número de cancelamentos de contratos dessa empresa, e para entregar uma solução para este problema. 


Para isso, a Umbrella Corp. forneceu um *dataset* com os detalhes dos cancelamentos. Os Pythonianos(🐍) devem analisar o arquivo utilizando o **Panda** (🐼).


Vamos começar trazendo todos os dados de *dataset*:

vamos começar trazendo todos os dados do dataset:

```python
import pandas as pd

tabela = pd.read_csv("cancelamentos.csv")
display(tabela)
```

## Tratamento de dados


Ao ler o *dataset*, percebemos que existem células com valore nulos, que atrapalham a nossa análise, e portanto devem ser retirados da análise.


Vamos Eliminá-los e re-exibir os valores, mas primeiro, precisaremos eliminar a coluna **ID** e re-exibir a tabela:

```python
tabela = tabela.drop("CustomerID", axis=1)
display(tabela)
```

Vamos pedir para que o Python nos retorne o número de linhas preenchidas da tabela.

```python
display(tabela.info())
```

Veja que temoc em cada uma das colunas um número deiferente de linhas. Isso acontece porque temos dados nulos na tabela, e que devemod eliminar. Portanto , vamos elimná-los e através do comando Python abaixo e re-exibir a tabela

```python
tebela = tabela.dropna()
display(tabela.info())
```

## Análise de dados


Ahora sim é que começa a análise de dados propriamente dita. Vamos começar verificando o número de clientes que cancelaram o contarto eo número de clientes que cancelaram.

```python
display(tabela["cancelou"].value_counts())
```

Podemos perceber que o número de clientes que cancelaram (1.0) realmente é alto, inclusive mais alto que o numero de clientes que não cancelaram, mas quanto que isso dá em procentagem? Vamos descobrir:

```python
display(tabela["cancelou"].value_counts(normalize=True).map("{:.2%}".format))
```

Sabemos agora que aproximadamente % dos clientes cancelaram seus contratos. Nosso trabalho é descobrit o porquê.

```python

display(tabela["duracao_contrato"].value_counts())
display(tabela["duracao_contrato"].value_counts(normalize=True).map("{:.2%}".format))
```

Podemos perceber que existe uma minoria em relação aos contratos mensais, o que talvez possa significar que os cancelamentos estejam vindo dos contratos mensais.

O que acontece se retirarmos da análise os contratos mensais? Vamos ver se isso resolve o problema da empresa:

```python
display(tabela["duracao_contrato"].value_counts())
display(tabela["duracao_contrato"].value_counts())


display(tabela.groupby("duracao_contrato").mean(numeric_only=True))
```

Logo de cara percebemoc que todos os clientes que optaram pelo tipo de contrato mensal cancelaram o serviço

o que pode ser feito para resolver o problema dos contrartos mensais.


Podemos retirar o tipo de contrato mensal da análise para verificar qual porcentagem de cancelamentos que cobra:

```python
tabel= tebela[tabela["duracao_contrato"] != "Momthly"]
display(tabela)
display(tabela["cancelou"].value_counts())
display(tabela["cancelou"].value_counts(normalize=True).map("{:2%}".format))
```

Podemos verificar que sem o contrato do tipo mensal, a taxa de cancelamentos diminuiu para 46%. Só que a taxa de cancelamentos continua muito alta. Poratnto , precisaremos fazer uma análise ainda mais profunda nos contratos do tipo anual e trimestra.


Par termos uma noção melhor , vamos criar gráfico de cada coluna para verificar se há discrepâncias muito grandes nas taxas de cancelamentos em algumas dessas colunas:

```python
import plotly.express as px

for coluna in tabela.columns:
    grafico = px.histogram(tabela, x=coluna, color="cancelou", width=600)

    grafico.show()
```

Os gráficos indicam problemas com os usuários que fazem mais que 5 ligaçãoes para o call centaer, e também cancelamentos por parte dos usuários que passam mais de 20 dias inadiplentes. Portanto,devemos eliminar esses clientes da análise.

```python
tabela = tabela[tabela["ligacoes_callcenter"] > 5]
tabela = tabela[tabela["dias_atraso"] <= 20]
display(tabela)
display(tabela["cancelou"].value_counts())
display(tabela["cancelou"].value_counts(normalize=True).map("{:.2%}".format))
```
