# Análise de dataset


A empresa Umbrella Corporation contratou desenvolvedores Python (🐍) para analisar um dataset para verificar o alto número de cancelamentos de contratos dessa empresa, e para entregar uma solução para este problema. 


Para isso, a Umbrella Corp. forneceu um *dataset* com os detalhes dos cancelamentos. Os Pythonianos(🐍) devem analisar o arquivo utilizando o **Panda** (🐼).


Vamos começar trazendo todos os dados de *dataset*:

vamos começar trazendo todos os dados do dataset:

```python
import pandas as pd

tabela = pd.read_csv("cancelamentos.csv")
display(tabela)
```

## Tratamento de dados


Ao ler o *dataset*, percebemos que existem células com valore nulos, que atrapalham a nossa análise, e portanto devem ser retirados da análise.


Vamos Eliminá-los e re-exibir os valores, mas primeiro, precisaremos eliminar a coluna **ID** e re-exibir a tabela:

```python
tabela = tabela.drop("CustomerID", axis=1)
display(tabela)
```

Vamos pedir para que o Python nos retorne o número de linhas preenchidas da tabela.

```python
display(tabela.info())
```

Veja que temoc em cada uma das colunas um número deiferente de linhas. Isso acontece porque temos dados nulos na tabela, e que devemod eliminar. Portanto , vamos elimná-los e através do comando Python abaixo e re-exibir a tabela

```python
tebela = tabela.dropna()
display(tabela.info())
```

## Análise de dados


Ahora sim é que começa a análise de dados propriamente dita. Vamos começar verificando o número de clientes que cancelaram o contarto eo número de clientes que cancelaram.

```python
display(tabela["cancelou"].value_counts())
```

Podemos perceber que o número de clientes que cancelaram (1.0) realmente é alto, inclusive mais alto que o numero de clientes que não cancelaram, mas quanto que isso dá em procentagem? Vamos descobrir:

```python
display(tabela["cancelou"].value_counts(normalize=True).map("{:.2%}".format))
```
